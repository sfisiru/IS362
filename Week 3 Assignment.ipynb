My bicycle odometer shows cumulative miles, logged by hand at the end of each day. Below I build a pandas Series from those cumulative odometer readings, then use it to show how many miles I actually rode each day.

```python
import pandas as pd
```

## Cumulative miles 

This is the running total that the odometer shows. Day 1 read 55, Day 2 read 120 (55 + 65).

```python
day_labels = [f"Day {i}" for i in range(1, 8)]

cumulative_miles = pd.Series(
    [55, 120, 190, 250, 325, 405, 470],
    index=day_labels,
    name="Cumulative Miles"
)
cumulative_miles
```

## Miles ridden each day

The odometer never resets, so the miles ridden on a given day is just that day's reading minus the day before's. I fill in Day 1 by hand since there's no prior reading to subtract.

```python
daily_miles = cumulative_miles.diff()
daily_miles.iloc[0] = cumulative_miles.iloc[0]
daily_miles = daily_miles.astype(int)
daily_miles.name = "Miles Ridden"
daily_miles
```
